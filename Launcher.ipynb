```python
import pandas as pd

# 1. Récupérer les fichiers Excel
df_qualification = pd.read_excel('CL.xlsx', sheet_name='Feuil1')
df_mise_en_service = pd.read_excel('Mes.xlsx', sheet_name='Feuil1')

# 2. Fusionner les données pour obtenir la production du jour
df_final = pd.merge(
    df_qualification, 
    df_mise_en_service[["Centres", "Production"]], 
    on="Centres", 
    how="left"
).rename(columns={"Production": "Production du jour"})

# 3. Calculer le total général de la production du jour
total_general = df_final["Total général"]

# 4. Calculer le taux de réalisation pour chaque centre
# Formule : (Production du jour de chaque centre / Total général) * 100
df_final["Taux de réalisation (%)"] = (df_final["Production du jour"] / total_general) * 100
df_final["Taux de réalisation (%)"]=df_final["Taux de réalisation (%)"].round(2)

# 5. Arrondir le taux à 2 décimales pour une meilleure lisibilité
df_final["Taux de réalisation (%)"] = df_final["Taux de réalisation (%)"]

# 6. Afficher le résultat
print("Total général de production :", total_general)
print("\nRésultats par centre :")
print(df_final)

# Export to Excel
df_final.to_excel('output_test.xlsx', index=False, sheet_name='Users')

# # 7. Optionnel : Afficher la somme des taux pour vérification (doit être ~100%)
# print(f"\nSomme des taux : {df_final['Taux de réalisation (%)'].sum():.2f}%")
```
